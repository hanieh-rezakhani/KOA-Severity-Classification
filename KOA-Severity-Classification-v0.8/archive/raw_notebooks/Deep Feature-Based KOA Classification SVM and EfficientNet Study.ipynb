```python
# ============================================================
# CELL 5 — 448x448 PREPROCESSING + DATALOADERS
# ============================================================

import random
import numpy as np
import cv2
import torch

from PIL import Image
from torchvision import transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader, WeightedRandomSampler


# ============================================================
# 1. Reproducibility
# ============================================================

SEED = 2026

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)


# ============================================================
# 2. CLAHE transform
# ============================================================

class CLAHETransform:

    def __init__(
        self,
        clip_limit=2.0,
        grid_size=(8, 8)
    ):
        self.clip_limit = clip_limit
        self.grid_size = grid_size

    def __call__(
        self,
        image
    ):

        # PIL -> NumPy
        image = np.array(image)

        # RGB -> grayscale
        if len(image.shape) == 3:

            image = cv2.cvtColor(
                image,
                cv2.COLOR_RGB2GRAY
            )

        # CLAHE
        clahe = cv2.createCLAHE(
            clipLimit=self.clip_limit,
            tileGridSize=self.grid_size
        )

        image = clahe.apply(
            image
        )

        # grayscale -> 3-channel RGB
        image = cv2.cvtColor(
            image,
            cv2.COLOR_GRAY2RGB
        )

        return Image.fromarray(
            image
        )


# ============================================================
# 3. Training transform — 448x448
# ============================================================

train_transform_448 = transforms.Compose([

    transforms.Resize(
        (448, 448)
    ),

    CLAHETransform(
        clip_limit=2.0,
        grid_size=(8, 8)
    ),

    transforms.RandomHorizontalFlip(
        p=0.5
    ),

    transforms.RandomRotation(
        degrees=5
    ),

    transforms.RandomAffine(
        degrees=0,
        translate=(0.03, 0.03),
        scale=(0.95, 1.05)
    ),

    transforms.ColorJitter(
        brightness=0.10,
        contrast=0.10
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],
        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])


# ============================================================
# 4. Validation / test transform — 448x448
# ============================================================

val_test_transform_448 = transforms.Compose([

    transforms.Resize(
        (448, 448)
    ),

    CLAHETransform(
        clip_limit=2.0,
        grid_size=(8, 8)
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[
            0.485,
            0.456,
            0.406
        ],
        std=[
            0.229,
            0.224,
            0.225
        ]
    )
])


# ============================================================
# 5. Create transformed datasets
# ============================================================

train_dataset_448 = ImageFolder(
    root=str(DATASET_PATH / "train"),
    transform=train_transform_448
)

val_dataset_448 = ImageFolder(
    root=str(DATASET_PATH / "val"),
    transform=val_test_transform_448
)

test_dataset_448 = ImageFolder(
    root=str(DATASET_PATH / "test"),
    transform=val_test_transform_448
)


# ============================================================
# 6. WeightedRandomSampler
# ============================================================

train_labels_448 = np.array(
    train_dataset_448.targets
)

class_counts_448 = np.bincount(
    train_labels_448,
    minlength=5
)

class_weights_448 = (
    1.0 / class_counts_448
)

sample_weights_448 = (
    class_weights_448[
        train_labels_448
    ]
)

sampler_generator_448 = (
    torch.Generator()
)

sampler_generator_448.manual_seed(
    SEED
)

sampler_448 = WeightedRandomSampler(
    weights=torch.as_tensor(
        sample_weights_448,
        dtype=torch.double
    ),
    num_samples=len(
        sample_weights_448
    ),
    replacement=True,
    generator=sampler_generator_448
)


# ============================================================
# 7. DataLoaders
# ============================================================

train_loader_448 = DataLoader(
    train_dataset_448,
    batch_size=16,
    sampler=sampler_448,
    num_workers=0,
    pin_memory=False
)

val_loader_448 = DataLoader(
    val_dataset_448,
    batch_size=16,
    shuffle=False,
    num_workers=0,
    pin_memory=False
)

test_loader_448 = DataLoader(
    test_dataset_448,
    batch_size=16,
    shuffle=False,
    num_workers=0,
    pin_memory=False
)


# ============================================================
# 8. Get one batch
# ============================================================

images_448, labels_448 = next(
    iter(train_loader_448)
)


# ============================================================
# 9. Verification
# ============================================================

print("=" * 70)
print("448x448 PIPELINE CHECK")
print("=" * 70)

print(
    "\nTrain samples:",
    len(train_dataset_448)
)

print(
    "Validation samples:",
    len(val_dataset_448)
)

print(
    "Test samples:",
    len(test_dataset_448)
)

print(
    "\nBatch image shape:",
    images_448.shape
)

print(
    "Batch label shape:",
    labels_448.shape
)

print(
    "\nClass mapping:",
    train_dataset_448.class_to_idx
)

print(
    "\nTrain class counts:",
    class_counts_448.tolist()
)

print(
    "\nPixel tensor range:"
)

print(
    "Min:",
    images_448.min().item()
)

print(
    "Max:",
    images_448.max().item()
)

# ============================================================
# 10. Assertions
# ============================================================

assert len(train_dataset_448) == 5778
assert len(val_dataset_448) == 826
assert len(test_dataset_448) == 1656

assert images_448.shape == (
    16,
    3,
    448,
    448
)

assert labels_448.shape == (
    16,
)

assert train_dataset_448.class_to_idx == {
    "0": 0,
    "1": 1,
    "2": 2,
    "3": 3,
    "4": 4
}

print("\n" + "=" * 70)
print("448x448 PIPELINE CHECK PASSED")
print("=" * 70)
```

```python
# ============================================================
# CELL 6 — LOAD TRAINED 448x448 RESNET18
# ============================================================

import torch
import torch.nn as nn
from pathlib import Path
from torchvision import models


# ============================================================
# 1. CPU device
# ============================================================

device = torch.device(
    "cpu"
)

print("=" * 70)
print("LOADING TRAINED 448x448 MODEL")
print("=" * 70)

print(
    "Device:",
    device
)


# ============================================================
# 2. Check checkpoint
# ============================================================

CHECKPOINT_PATH = (
    Path.cwd()
    /
    "best_resnet18_448_seed2026.pth"
)

print(
    "\nCheckpoint:"
)

print(
    CHECKPOINT_PATH
)

print(
    "Exists:",
    CHECKPOINT_PATH.exists()
)


if not CHECKPOINT_PATH.exists():

    raise FileNotFoundError(
        "\nCheckpoint not found.\n\n"
        "Put this file in the main PyCharm project folder:\n"
        "best_resnet18_448_seed2026.pth"
    )


# ============================================================
# 3. Create ResNet18 architecture
# ============================================================

model_448 = models.resnet18(
    weights=None
)


# Replace final layer with 5 classes
model_448.fc = nn.Linear(
    model_448.fc.in_features,
    5
)


# ============================================================
# 4. Load checkpoint
# ============================================================

checkpoint = torch.load(
    CHECKPOINT_PATH,
    map_location=device,
    weights_only=False
)


print(
    "\nCheckpoint keys:"
)

print(
    list(checkpoint.keys())
)


# Load trained weights
model_448.load_state_dict(
    checkpoint[
        "model_state_dict"
    ]
)


# Move to CPU
model_448 = model_448.to(
    device
)

model_448.eval()


# ============================================================
# 5. Display stored training information
# ============================================================

print("\n" + "=" * 70)
print("STORED CHECKPOINT INFORMATION")
print("=" * 70)

print(
    "Best epoch:",
    checkpoint.get(
        "best_epoch"
    )
)

print(
    "Macro F1:",
    checkpoint.get(
        "macro_f1"
    )
)

print(
    "Balanced Accuracy:",
    checkpoint.get(
        "balanced_acc"
    )
)

print(
    "Accuracy:",
    checkpoint.get(
        "accuracy"
    )
)

print(
    "MAE:",
    checkpoint.get(
        "mae"
    )
)

print(
    "QWK:",
    checkpoint.get(
        "qwk"
    )
)

print(
    "Seed:",
    checkpoint.get(
        "seed"
    )
)

print(
    "Resolution:",
    checkpoint.get(
        "resolution"
    )
)


# ============================================================
# 6. Test forward pass
# ============================================================

images, labels = next(
    iter(val_loader_448)
)

images = images.to(
    device
)

with torch.no_grad():

    outputs = model_448(
        images
    )

predictions = torch.argmax(
    outputs,
    dim=1
)


# ============================================================
# 7. Verification
# ============================================================

print("\n" + "=" * 70)
print("FORWARD PASS CHECK")
print("=" * 70)

print(
    "Input shape:",
    images.shape
)

print(
    "Output shape:",
    outputs.shape
)

print(
    "Predictions shape:",
    predictions.shape
)

print(
    "First 10 true labels:",
    labels[:10].tolist()
)

print(
    "First 10 predictions:",
    predictions[:10].tolist()
)

assert outputs.shape == (
    16,
    5
)

assert predictions.shape == (
    16,
)

print("\n" + "=" * 70)
print("448x448 CHECKPOINT LOAD PASSED")
print("=" * 70)
```

```python
# ============================================================
# CELL 7 — EXTRACT 512-D DEEP FEATURES
# 448x448 TRAIN + VALIDATION
# CPU-ONLY
# ============================================================

import time
import numpy as np
import torch
import torch.nn as nn

from pathlib import Path
from tqdm.auto import tqdm
from torchvision.datasets import ImageFolder
from torch.utils.data import DataLoader


# ============================================================
# 1. Device
# ============================================================

device = torch.device("cpu")


# ============================================================
# 2. Create deterministic datasets
#
# IMPORTANT:
# No random augmentation during feature extraction.
# ============================================================

feature_train_dataset = ImageFolder(
    root=str(
        DATASET_PATH / "train"
    ),
    transform=val_test_transform_448
)

feature_val_dataset = ImageFolder(
    root=str(
        DATASET_PATH / "val"
    ),
    transform=val_test_transform_448
)


# ============================================================
# 3. DataLoaders
#
# We do NOT use WeightedRandomSampler here.
# Every image should be processed exactly once.
# ============================================================

FEATURE_BATCH_SIZE = 8

feature_train_loader = DataLoader(
    feature_train_dataset,
    batch_size=FEATURE_BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=False
)

feature_val_loader = DataLoader(
    feature_val_dataset,
    batch_size=FEATURE_BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=False
)


# ============================================================
# 4. Turn ResNet18 into a feature extractor
#
# Original:
#
# image → ... → 512 features → FC → 5 classes
#
# We remove FC:
#
# image → ... → 512 features
# ============================================================

feature_extractor = nn.Sequential(
    *list(
        model_448.children()
    )[:-1]
)

feature_extractor = (
    feature_extractor.to(device)
)

feature_extractor.eval()


# ============================================================
# 5. Feature extraction function
# ============================================================

def extract_features(
    model,
    loader,
    split_name
):

    all_features = []
    all_labels = []

    start_time = time.time()

    print(
        "\n" + "=" * 70
    )

    print(
        f"EXTRACTING {split_name.upper()} FEATURES"
    )

    print(
        "=" * 70
    )

    with torch.no_grad():

        for images, labels in tqdm(
            loader,
            desc=f"{split_name} batches"
        ):

            images = images.to(
                device
            )

            # Forward through ResNet
            features = model(
                images
            )

            # Current shape:
            # [batch, 512, 1, 1]

            features = features.flatten(
                start_dim=1
            )

            # Final shape:
            # [batch, 512]

            all_features.append(
                features.cpu().numpy()
            )

            all_labels.append(
                labels.numpy()
            )


    # Combine all batches
    features_array = np.concatenate(
        all_features,
        axis=0
    )

    labels_array = np.concatenate(
        all_labels,
        axis=0
    )

    elapsed = (
        time.time()
        - start_time
    )

    print(
        "\nFeature extraction complete."
    )

    print(
        "Feature shape:",
        features_array.shape
    )

    print(
        "Label shape:",
        labels_array.shape
    )

    print(
        "Time:",
        round(elapsed, 2),
        "seconds"
    )

    return (
        features_array,
        labels_array
    )


# ============================================================
# 6. Extract Train features
# ============================================================

X_train_deep, y_train_deep = (
    extract_features(
        feature_extractor,
        feature_train_loader,
        "train"
    )
)


# ============================================================
# 7. Extract Validation features
# ============================================================

X_val_deep, y_val_deep = (
    extract_features(
        feature_extractor,
        feature_val_loader,
        "validation"
    )
)


# ============================================================
# 8. Verification
# ============================================================

print(
    "\n" + "=" * 70
)

print(
    "FEATURE EXTRACTION CHECK"
)

print(
    "=" * 70
)

print(
    "Train features:",
    X_train_deep.shape
)

print(
    "Train labels:",
    y_train_deep.shape
)

print(
    "Validation features:",
    X_val_deep.shape
)

print(
    "Validation labels:",
    y_val_deep.shape
)


assert X_train_deep.shape == (
    5778,
    512
)

assert y_train_deep.shape == (
    5778,
)

assert X_val_deep.shape == (
    826,
    512
)

assert y_val_deep.shape == (
    826,
)


# ============================================================
# 9. Check class distributions
# ============================================================

print("\nTrain class counts:")

print(
    np.bincount(
        y_train_deep,
        minlength=5
    )
)

print(
    "\nValidation class counts:"
)

print(
    np.bincount(
        y_val_deep,
        minlength=5
    )
)


# ============================================================
# 10. Save features
# ============================================================

train_feature_path = (
    Path.cwd()
    /
    "train_deep_features_448_seed2026.npz"
)

val_feature_path = (
    Path.cwd()
    /
    "val_deep_features_448_seed2026.npz"
)


np.savez_compressed(
    train_feature_path,
    features=X_train_deep,
    labels=y_train_deep
)

np.savez_compressed(
    val_feature_path,
    features=X_val_deep,
    labels=y_val_deep
)


print(
    "\n" + "=" * 70
)

print(
    "FEATURES SAVED"
)

print(
    "=" * 70
)

print(
    "Train:"
)

print(
    train_feature_path
)

print(
    "Validation:"
)

print(
    val_feature_path
)

print(
    "\n" + "=" * 70
)

print(
    "512-D FEATURE EXTRACTION PASSED"
)

print(
    "=" * 70
)
```

```python
# ============================================================
# CELL 8 — 512-D DEEP FEATURES + RBF SVM
# NO DIMENSIONALITY REDUCTION
# ============================================================

import time
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    mean_absolute_error,
    cohen_kappa_score,
    confusion_matrix
)


# ============================================================
# 1. Load saved features
# ============================================================

print("=" * 70)
print("LOADING 512-D FEATURES")
print("=" * 70)

train_data = np.load(
    "train_deep_features_448_seed2026.npz"
)

val_data = np.load(
    "val_deep_features_448_seed2026.npz"
)

X_train = train_data["features"]
y_train = train_data["labels"]

X_val = val_data["features"]
y_val = val_data["labels"]


print(
    "Train:",
    X_train.shape
)

print(
    "Validation:",
    X_val.shape
)


# ============================================================
# 2. Standardization
#
# IMPORTANT:
# fit ONLY on training data
# ============================================================

print("\n" + "=" * 70)
print("STANDARDIZING FEATURES")
print("=" * 70)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_val_scaled = scaler.transform(
    X_val
)

print(
    "Standardization complete."
)


# ============================================================
# 3. Create RBF SVM
# ============================================================

print("\n" + "=" * 70)
print("TRAINING 512-D RBF SVM")
print("=" * 70)

print(
    "Kernel: RBF"
)

print(
    "C: 1.0"
)

print(
    "Gamma: scale"
)


svm_512 = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale",
    cache_size=4000
)


# ============================================================
# 4. Train SVM
# ============================================================

start_time = time.time()

svm_512.fit(
    X_train_scaled,
    y_train
)

training_time = (
    time.time()
    - start_time
)


print(
    "\nSVM training completed."
)

print(
    "Training time:",
    round(
        training_time,
        2
    ),
    "seconds"
)


# ============================================================
# 5. Predict validation set
# ============================================================

print(
    "\nPredicting validation set..."
)

start_time = time.time()

y_pred = svm_512.predict(
    X_val_scaled
)

prediction_time = (
    time.time()
    - start_time
)

print(
    "Prediction time:",
    round(
        prediction_time,
        2
    ),
    "seconds"
)


# ============================================================
# 6. Metrics
# ============================================================

accuracy = accuracy_score(
    y_val,
    y_pred
)

balanced_acc = (
    balanced_accuracy_score(
        y_val,
        y_pred
    )
)

macro_f1 = f1_score(
    y_val,
    y_pred,
    average="macro",
    zero_division=0
)

mae = mean_absolute_error(
    y_val,
    y_pred
)

qwk = cohen_kappa_score(
    y_val,
    y_pred,
    weights="quadratic"
)

adjacent_acc = np.mean(
    np.abs(
        y_val - y_pred
    ) <= 1
)

per_class_f1 = f1_score(
    y_val,
    y_pred,
    labels=[
        0,
        1,
        2,
        3,
        4
    ],
    average=None,
    zero_division=0
)

cm = confusion_matrix(
    y_val,
    y_pred,
    labels=[
        0,
        1,
        2,
        3,
        4
    ]
)


# ============================================================
# 7. Results
# ============================================================

print(
    "\n" + "=" * 70
)

print(
    "512-D SVM VALIDATION RESULT"
)

print(
    "=" * 70
)

print(
    f"Accuracy         : {accuracy:.4f}"
)

print(
    f"Balanced Accuracy: {balanced_acc:.4f}"
)

print(
    f"Macro F1         : {macro_f1:.4f}"
)

print(
    f"MAE              : {mae:.4f}"
)

print(
    f"QWK              : {qwk:.4f}"
)

print(
    f"Adjacent Accuracy: {adjacent_acc:.4f}"
)


print(
    "\nPer-class F1:"
)

for class_id in range(5):

    print(
        f"KL{class_id}: "
        f"{per_class_f1[class_id]:.4f}"
    )


print(
    "\nConfusion matrix:"
)

print(
    cm
)


# ============================================================
# 8. Save baseline results
# ============================================================

svm_result = pd.DataFrame([{

    "Method":
        "512D + RBF SVM",

    "Accuracy":
        accuracy,

    "Balanced_Accuracy":
        balanced_acc,

    "Macro_F1":
        macro_f1,

    "MAE":
        mae,

    "QWK":
        qwk,

    "Adjacent_Accuracy":
        adjacent_acc,

    "KL0_F1":
        per_class_f1[0],

    "KL1_F1":
        per_class_f1[1],

    "KL2_F1":
        per_class_f1[2],

    "KL3_F1":
        per_class_f1[3],

    "KL4_F1":
        per_class_f1[4]

}])


svm_result.to_csv(
    "svm_512_baseline_448.csv",
    index=False
)


print(
    "\nResults saved to:"
)

print(
    "svm_512_baseline_448.csv"
)
```

```python
# ============================================================
# CELL 2 — TRAIN EFFICIENTNET-B0 ON 448x448 X-RAYS
# ============================================================

import time
import random
import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    mean_absolute_error,
    cohen_kappa_score,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. Reproducibility
# ------------------------------------------------------------
SEED = 2026

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

print("=" * 70)
print("TRAINING EFFICIENTNET-B0")
print("=" * 70)

print(f"Seed       : {SEED}")
print(f"Device     : {device}")
print(f"Epochs     : 10")
print(f"Learning rate: 0.0001")
print()

# ------------------------------------------------------------
# 2. Loss and optimizer
# ------------------------------------------------------------
criterion = nn.CrossEntropyLoss()

optimizer = torch.optim.Adam(
    model_effnet.parameters(),
    lr=1e-4,
    weight_decay=0
)

# ------------------------------------------------------------
# 3. Helper function for validation
# ------------------------------------------------------------
def evaluate_model(model, loader, criterion, device):

    model.eval()

    total_loss = 0.0
    all_preds = []
    all_labels = []

    with torch.no_grad():

        for images, labels in loader:

            images = images.to(device)
            labels = labels.to(device)

            outputs = model(images)

            loss = criterion(
                outputs,
                labels
            )

            total_loss += (
                loss.item() * images.size(0)
            )

            preds = torch.argmax(
                outputs,
                dim=1
            )

            all_preds.extend(
                preds.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

    avg_loss = (
        total_loss / len(loader.dataset)
    )

    all_preds = np.array(all_preds)
    all_labels = np.array(all_labels)

    accuracy = accuracy_score(
        all_labels,
        all_preds
    )

    balanced_acc = balanced_accuracy_score(
        all_labels,
        all_preds
    )

    macro_f1 = f1_score(
        all_labels,
        all_preds,
        average="macro"
    )

    mae = mean_absolute_error(
        all_labels,
        all_preds
    )

    qwk = cohen_kappa_score(
        all_labels,
        all_preds,
        weights="quadratic"
    )

    adjacent_accuracy = np.mean(
        np.abs(all_labels - all_preds) <= 1
    )

    return {
        "loss": avg_loss,
        "accuracy": accuracy,
        "balanced_accuracy": balanced_acc,
        "macro_f1": macro_f1,
        "mae": mae,
        "qwk": qwk,
        "adjacent_accuracy": adjacent_accuracy,
        "preds": all_preds,
        "labels": all_labels
    }


# ------------------------------------------------------------
# 4. Training loop
# ------------------------------------------------------------
NUM_EPOCHS = 10

best_macro_f1 = -np.inf
best_epoch = None
best_state_dict = None
best_metrics = None

history = []

overall_start = time.time()

for epoch in range(1, NUM_EPOCHS + 1):

    epoch_start = time.time()

    model_effnet.train()

    running_loss = 0.0

    for images, labels in train_loader:

        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        outputs = model_effnet(images)

        loss = criterion(
            outputs,
            labels
        )

        loss.backward()

        optimizer.step()

        running_loss += (
            loss.item() * images.size(0)
        )

    train_loss = (
        running_loss / len(train_loader.dataset)
    )

    # --------------------------------------------------------
    # Validation
    # --------------------------------------------------------
    val_metrics = evaluate_model(
        model_effnet,
        val_loader,
        criterion,
        device
    )

    epoch_time = time.time() - epoch_start

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "val_loss": val_metrics["loss"],
        "val_accuracy": val_metrics["accuracy"],
        "val_balanced_accuracy": val_metrics["balanced_accuracy"],
        "val_macro_f1": val_metrics["macro_f1"],
        "val_mae": val_metrics["mae"],
        "val_qwk": val_metrics["qwk"],
        "val_adjacent_accuracy": val_metrics["adjacent_accuracy"]
    })

    print(
        f"Epoch {epoch:02d}/{NUM_EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {val_metrics['loss']:.4f} | "
        f"Val Acc: {val_metrics['accuracy']:.4f} | "
        f"Val Macro F1: {val_metrics['macro_f1']:.4f} | "
        f"Val QWK: {val_metrics['qwk']:.4f} | "
        f"Time: {epoch_time:.1f}s"
    )

    # --------------------------------------------------------
    # Save best model based on validation Macro F1
    # --------------------------------------------------------
    if val_metrics["macro_f1"] > best_macro_f1:

        best_macro_f1 = val_metrics["macro_f1"]
        best_epoch = epoch

        best_state_dict = {
            k: v.detach().cpu().clone()
            for k, v in model_effnet.state_dict().items()
        }

        best_metrics = val_metrics.copy()


overall_time = time.time() - overall_start

# ------------------------------------------------------------
# 5. Restore best model
# ------------------------------------------------------------
model_effnet.load_state_dict(
    best_state_dict
)

model_effnet = model_effnet.to(device)

# ------------------------------------------------------------
# 6. Final best validation evaluation
# ------------------------------------------------------------
final_val_metrics = evaluate_model(
    model_effnet,
    val_loader,
    criterion,
    device
)

print()
print("=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

print(f"Total training time: {overall_time:.2f} sec")
print(f"Best epoch         : {best_epoch}")

print()
print("BEST VALIDATION RESULT")
print(f"Accuracy         : {final_val_metrics['accuracy']:.4f}")
print(f"Balanced Accuracy: {final_val_metrics['balanced_accuracy']:.4f}")
print(f"Macro F1         : {final_val_metrics['macro_f1']:.4f}")
print(f"MAE              : {final_val_metrics['mae']:.4f}")
print(f"QWK              : {final_val_metrics['qwk']:.4f}")
print(
    f"Adjacent Accuracy: "
    f"{final_val_metrics['adjacent_accuracy']:.4f}"
)

# ------------------------------------------------------------
# 7. Per-class F1
# ------------------------------------------------------------
per_class_f1 = f1_score(
    final_val_metrics["labels"],
    final_val_metrics["preds"],
    average=None,
    labels=[0, 1, 2, 3, 4]
)

print()
print("Per-class F1:")

for class_id, score in enumerate(per_class_f1):
    print(f"KL{class_id}: {score:.4f}")

# ------------------------------------------------------------
# 8. Confusion matrix
# ------------------------------------------------------------
cm = confusion_matrix(
    final_val_metrics["labels"],
    final_val_metrics["preds"],
    labels=[0, 1, 2, 3, 4]
)

print()
print("Confusion matrix:")
print(cm)

# ------------------------------------------------------------
# 9. Save checkpoint
# ------------------------------------------------------------
checkpoint = {
    "model_name": "EfficientNet-B0",
    "num_classes": 5,
    "resolution": 448,
    "seed": SEED,
    "best_epoch": best_epoch,
    "macro_f1": final_val_metrics["macro_f1"],
    "accuracy": final_val_metrics["accuracy"],
    "balanced_accuracy": final_val_metrics["balanced_accuracy"],
    "mae": final_val_metrics["mae"],
    "qwk": final_val_metrics["qwk"],
    "adjacent_accuracy": final_val_metrics["adjacent_accuracy"],
    "model_state_dict": model_effnet.state_dict()
}

torch.save(
    checkpoint,
    "best_efficientnetb0_448_seed2026.pth"
)

# ------------------------------------------------------------
# 10. Save training history
# ------------------------------------------------------------
history_df = pd.DataFrame(history)

history_df.to_csv(
    "efficientnetb0_448_training_history.csv",
    index=False
)

print()
print("Saved:")
print("best_efficientnetb0_448_seed2026.pth")
print("efficientnetb0_448_training_history.csv")
```
