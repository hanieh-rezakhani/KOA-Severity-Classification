```python

import docx
from docx.shared import Inches, Pt, RGBColor
from docx.enum.text import WD_ALIGN_PARAGRAPH
from docx.enum.table import WD_TABLE_ALIGNMENT, WD_ALIGN_VERTICAL
from docx.oxml import OxmlElement, parse_xml
from docx.oxml.ns import nsdecls, qn

def create_proposal_document():
    doc = docx.Document()

    # --- Page Setup ---
    for section in doc.sections:
        section.top_margin = Inches(1)
        section.bottom_margin = Inches(1)
        section.left_margin = Inches(1)
        section.right_margin = Inches(1)

    # --- Color Palette ---
    COLOR_PRIMARY = RGBColor(31, 78, 121)    # Deep Navy
    COLOR_SECONDARY = RGBColor(89, 89, 89)   # Slate Grey
    COLOR_TEXT = RGBColor(38, 38, 38)        # Charcoal Text
    HEX_PRIMARY = "1F4E79"
    HEX_LIGHT_BG = "F2F4F7"
    HEX_BORDER = "D3D3D3"

    # --- Helper Functions ---
    def add_title(text):
        p = doc.add_paragraph()
        p.alignment = WD_ALIGN_PARAGRAPH.CENTER
        p.paragraph_format.space_before = Pt(0)
        p.paragraph_format.space_after = Pt(12)
        run = p.add_run(text)
        run.font.name = 'Calibri'
        run.font.size = Pt(22)
        run.font.bold = True
        run.font.color.rgb = COLOR_PRIMARY

    def add_heading_1(text):
        p = doc.add_paragraph()
        p.paragraph_format.space_before = Pt(16)
        p.paragraph_format.space_after = Pt(6)
        p.paragraph_format.keep_with_next = True
        run = p.add_run(text)
        run.font.name = 'Calibri'
        run.font.size = Pt(14)
        run.font.bold = True
        run.font.color.rgb = COLOR_PRIMARY

    def add_heading_2(text):
        p = doc.add_paragraph()
        p.paragraph_format.space_before = Pt(12)
        p.paragraph_format.space_after = Pt(4)
        p.paragraph_format.keep_with_next = True
        run = p.add_run(text)
        run.font.name = 'Calibri'
        run.font.size = Pt(12)
        run.font.bold = True
        run.font.color.rgb = COLOR_SECONDARY

    def add_paragraph(text, bold_prefix="", italic=False):
        p = doc.add_paragraph()
        p.paragraph_format.space_before = Pt(0)
        p.paragraph_format.space_after = Pt(6)
        p.paragraph_format.line_spacing = 1.15

        if bold_prefix:
            r_bold = p.add_run(bold_prefix)
            r_bold.font.name = 'Calibri'
            r_bold.font.size = Pt(11)
            r_bold.font.bold = True
            r_bold.font.color.rgb = COLOR_TEXT

        run = p.add_run(text)
        run.font.name = 'Calibri'
        run.font.size = Pt(11)
        run.font.italic = italic
        run.font.color.rgb = COLOR_TEXT
        return p

    def add_bullet(text, bold_prefix=""):
        p = doc.add_paragraph(style='List Bullet')
        p.paragraph_format.space_before = Pt(0)
        p.paragraph_format.space_after = Pt(3)
        p.paragraph_format.line_spacing = 1.15

        if bold_prefix:
            r_bold = p.add_run(bold_prefix)
            r_bold.font.name = 'Calibri'
            r_bold.font.size = Pt(11)
            r_bold.font.bold = True
            r_bold.font.color.rgb = COLOR_TEXT

        run = p.add_run(text)
        run.font.name = 'Calibri'
        run.font.size = Pt(11)
        run.font.color.rgb = COLOR_TEXT

    def set_cell_background(cell, hex_color):
        shading = parse_xml(f'<w:shd {nsdecls("w")} w:fill="{hex_color}"/>')
        cell._tc.get_or_add_tcPr().append(shading)

    def set_cell_margins(cell, top=100, bottom=100, left=150, right=150):
        tcPr = cell._tc.get_or_add_tcPr()
        tcMar = OxmlElement('w:tcMar')
        for margin, value in [('top', top), ('bottom', bottom), ('left', left), ('right', right)]:
            node = OxmlElement(f'w:{margin}')
            node.set(qn('w:w'), str(value))
            node.set(qn('w:type'), 'dxa')
            tcMar.append(node)
        tcPr.append(tcMar)

    # --- DOCUMENT CONTENT ---

    # Document Title
    add_title("A Hybrid Deep Feature and Machine Learning Framework for Automated Knee Osteoarthritis Severity Grading from Radiographs")

    # Subtitle / Document Info
    p_meta = doc.add_paragraph()
    p_meta.alignment = WD_ALIGN_PARAGRAPH.CENTER
    p_meta.paragraph_format.space_after = Pt(18)
    r_meta = p_meta.add_run("Proposal & Methodological Benchmarks Report | File: KOA_Proposal_Polished_Benchmarks")
    r_meta.font.name = 'Calibri'
    r_meta.font.size = Pt(10)
    r_meta.font.italic = True
    r_meta.font.color.rgb = COLOR_SECONDARY

    doc.add_paragraph().paragraph_format.space_after = Pt(6)

    # Section 1
    add_heading_1("1. RESEARCH STATEMENT & OBJECTIVE")
    add_paragraph(
        "This project investigates whether combining frozen deep feature representations from pretrained convolutional "
        "neural networks with conventional machine learning classifiers and ordinality-inspired decision strategies "
        "can improve automated knee osteoarthritis (KOA) severity grading from radiographs, particularly under severe class imbalance conditions."
    )
    add_paragraph(
        "Here, 'Hybrid' refers specifically to the integration of deep feature representations with classical machine learning "
        "classifiers (e.g., Support Vector Machines, LightGBM) and probability-based decision strategies."
    )

    # Section 2
    add_heading_1("2. MOTIVATION & WHY CONVNEXT?")
    add_paragraph(
        "Initial exploration revealed that traditional end-to-end architectures (e.g., ResNet18) struggled significantly "
        "with minority classes (such as KL Grade 1 Doubtful and KL Grade 4 Severe) under severe class imbalance. Furthermore, "
        "while Vision Transformers (ViTs) offer strong global context aggregation, they suffer from hyperparameter sensitivity "
        "and high data appetite on constrained medical datasets."
    )
    add_paragraph("ConvNeXt was selected as the primary feature extractor due to two core factors:")
    add_bullet("It incorporates micro-design principles from Vision Transformers (e.g., 7x7 depthwise convolutions, LayerNorm) while retaining essential spatial inductive biases critical for radiograph interpretation.", "1. Modern CNN Design: ")
    add_bullet("It offers superior feature extraction stability without requiring full end-to-end retraining, avoiding overfitting on limited medical partitions.", "2. Transfer Efficacy: ")

    # Section 3
    add_heading_1("3. DATASET & LEAKAGE-AWARE PARTITIONING")
    add_paragraph(
        "The proposed framework was evaluated on a public knee osteoarthritis radiographic dataset categorized into "
        "five Kellgren-Lawrence (KL) severity grades (0–4). The evaluated dataset comprises 8,260 images:"
    )
    add_bullet("5,778 images (2,889 unique IDs)", "Training Set: ")
    add_bullet("826 images (413 unique IDs)", "Validation Set: ")
    add_bullet("1,656 images (828 unique IDs)", "Test Set: ")

    add_heading_2("Patient-Level Leakage Prevention")
    add_paragraph(
        "An additional auto_test partition (1,526 images) provided in raw source repositories was explicitly excluded "
        "due to verified ID overlap (811 overlapping IDs) with the test set. By strictly utilizing isolated image IDs "
        "across train, validation, and test splits, patient-level data leakage was completely prevented."
    )

    add_heading_2("Class Imbalance Distribution (Training Split)")
    add_bullet("2,286 images (39.56%)", "KL 0 (Normal): ")
    add_bullet("1,046 images (18.10%)", "KL 1 (Doubtful): ")
    add_bullet("1,516 images (26.24%)", "KL 2 (Mild): ")
    add_bullet("757 images (13.10%)", "KL 3 (Moderate): ")
    add_bullet("173 images (2.99%)", "KL 4 (Severe): ")

    # Section 4
    add_heading_1("4. PRELIMINARY EXPERIMENTS & METHODOLOGICAL EVOLUTION")

    add_heading_2("Phase 1: Deep Learning Baselines (ResNet18)")
    add_paragraph(
        "Standard ResNet18 baseline models trained end-to-end achieved a maximum Validation Accuracy of 63.56% and "
        "Macro F1 of 0.6560 (with Data Augmentation). However, severe performance bottlenecks were observed in borderline "
        "minority grades: KL 1 F1-score plateaued at ~0.3882, indicating that standard CNN backbones failed to resolve subtle joint-space narrowing features."
    )

    add_heading_2("Phase 2: Anatomical Landmark Investigation (KNEEL) & Domain Shift Analysis")
    add_paragraph(
        "To evaluate whether explicit anatomical localization could mitigate feature ambiguity, we integrated KNEEL "
        "(a pretrained landmark localization model)."
    )
    add_bullet(
        "When applied to single-knee cropped PNG radiographs, KNEEL exhibited severe Domain Shift. Because KNEEL "
        "was pre-trained on standardized clinical DICOM bilateral radiographs, landmark predictions collapsed (detecting dual knee structures on single knee images).",
        "Observation: "
    )
    add_bullet(
        "Quantitative anatomical landmark extraction was deemed unviable for direct KL grading on compressed single-knee images, "
        "motivating a shift toward robust deep visual embeddings.",
        "Conclusion: "
    )

    add_heading_2("Phase 3: Classical ML & Feature Engineering Exploration")
    add_paragraph(
        "Experiments with PCA dimensionality reduction combined with classical classifiers (SVM, LightGBM) demonstrated "
        "surprising resilience against noise and class imbalance, outperforming standard end-to-end neural baselines. "
        "This confirmed the utility of a hybrid feature-based pipeline."
    )

    # Section 5
    add_heading_1("5. PROPOSED METHODOLOGY & PIPELINE")
    add_paragraph("The finalized framework follows a strictly controlled, leakage-aware evaluation pipeline:")
    add_bullet("Frozen ConvNeXt-Small feature embeddings are extracted from input radiographs.", "1. Deep Visual Representation Extraction: ")
    add_bullet("Dimensionality reduction parameters are fitted exclusively on training features to eliminate information leakage.", "2. Training-Only PCA Compression: ")
    add_bullet("Model tuning is performed strictly using the validation split.", "3. Validation-Guided Hyperparameter Selection: ")
    add_bullet(
        "To reflect the natural progression of KOA (KL 0 < 1 < 2 < 3 < 4), the decision space is decomposed into two stages: "
        "Stage 1 (Binary Decomposition: Early-stage KL 0,1,2 vs Advanced-stage KL 3,4) and Stage 2 (Fine-Grained Classification: "
        "Probability-weighted Soft Ensemble combining Support Vector Machines and LightGBM).",
        "4. Ordinality-Inspired Hierarchical Decision Strategy: "
    )

    # Section 6
    add_heading_1("6. EXPERIMENTAL RESULTS & COMPARATIVE BENCHMARK")
    add_paragraph("Evaluated on the held-out test partition (1,656 images):")

    # Table Creation
    table = doc.add_table(rows=5, cols=5)
    table.alignment = WD_TABLE_ALIGNMENT.CENTER
    table.autofit = False

    col_widths = [Inches(2.0), Inches(0.8), Inches(0.9), Inches(1.0), Inches(0.9)]
    headers = ["Model / Strategy", "Test Acc", "Macro F1", "Balanced Acc", "Macro AUC"]

    # Format Header Row
    hdr_cells = table.rows[0].cells
    for i, title in enumerate(headers):
        hdr_cells[i].text = title
        set_cell_background(hdr_cells[i], HEX_PRIMARY)
        set_cell_margins(hdr_cells[i])
        hdr_cells[i].paragraphs[0].alignment = WD_ALIGN_PARAGRAPH.CENTER
        for run in hdr_cells[i].paragraphs[0].runs:
            run.font.name = 'Calibri'
            run.font.size = Pt(10)
            run.font.bold = True
            run.font.color.rgb = RGBColor(255, 255, 255)

    # Table Data
    data = [
        ["ResNet18 (Baseline + Aug)", "63.56%", "0.6560", "0.6669", "N/A"],
        ["ConvNeXt + Weighted Focal", "69.38%", "0.7234", "0.7313", "0.9139"],
        ["ConvNeXt + Ordinal Loss", "71.50%", "0.7192", "0.7101", "0.5672"],
        ["Proposed Hybrid (Ensemble)", "72.71%", "0.7239", "0.7380", "0.9268"]
    ]

    for row_idx, row_data in enumerate(data):
        row_cells = table.rows[row_idx + 1].cells
        bg_color = HEX_LIGHT_BG if row_idx % 2 == 1 else "FFFFFF"

        # Highlight proposed method
        is_proposed = row_idx == 3
        if is_proposed:
            bg_color = "E6F0FA"

        for col_idx, cell_value in enumerate(row_data):
            row_cells[col_idx].text = cell_value
            set_cell_background(row_cells[col_idx], bg_color)
            set_cell_margins(row_cells[col_idx])

            p = row_cells[col_idx].paragraphs[0]
            if col_idx > 0:
                p.alignment = WD_ALIGN_PARAGRAPH.CENTER

            for run in p.runs:
                run.font.name = 'Cambria'
                run.font.size = Pt(10)
                if is_proposed:
                    run.font.bold = True
                    run.font.color.rgb = COLOR_PRIMARY
                else:
                    run.font.color.rgb = COLOR_TEXT

    # Set Column Widths
    for row in table.rows:
        for i, w in enumerate(col_widths):
            row.cells[i].width = w

    doc.add_paragraph().paragraph_format.space_after = Pt(6)

    add_heading_2("Per-Class One-vs-Rest ROC-AUC (Proposed Framework)")
    add_bullet("0.9368", "KL 0 (Normal): ")
    add_bullet("0.7951", "KL 1 (Doubtful): ")
    add_bullet("0.9229", "KL 2 (Mild): ")
    add_bullet("0.9799", "KL 3 (Moderate): ")
    add_bullet("0.9982", "KL 4 (Severe): ")

    # Section 7
    add_heading_1("7. LIMITATIONS & FUTURE DIRECTIONS")
    add_bullet("Evaluation on single-source Kaggle dataset limits multi-center generalizability assessment.", "External Validation: ")
    add_bullet("Future work will incorporate SHAP and Grad-CAM for visual feature transparency.", "Explainability Integration: ")

    # Save File
    file_name = "KOA_Proposal_Polished_Benchmarks.docx"
    doc.save(file_name)
    print(f"✅ فایل با موفقیت ساخته شد: {file_name}")

create_proposal_document()
```

```python
import os
import glob
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torchvision import transforms
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from sklearn.metrics import accuracy_score, f1_score, cohen_kappa_score, mean_absolute_error, roc_auc_score, confusion_matrix
import timm

# ---------------------------------------------------------
# ۱. ساخت دیتالودر اختصاصی برای تصاویر دانلود شده
# ---------------------------------------------------------
class KOADataset(Dataset):
    def __init__(self, root_dir, transform=None):
        self.root_dir = root_dir
        self.transform = transform
        self.image_paths = []
        self.labels = []

        for label in range(5):
            class_dir = os.path.join(root_dir, str(label))
            if os.path.exists(class_dir):
                for img_name in os.listdir(class_dir):
                    if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                        self.image_paths.append(os.path.join(class_dir, img_name))
                        self.labels.append(label)

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        image = Image.open(img_path).convert('RGB')
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label

test_transform = transforms.Compose([
    transforms.Resize((384, 384)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

test_dir = os.path.join(path, "kneeKL22", "test")
if not os.path.exists(test_dir):
    possible_paths = glob.glob(os.path.join(path, "**", "test"), recursive=True)
    if possible_paths:
        test_dir = possible_paths[0]

print(f"📁 مسیر پوشه تست: {test_dir}")

test_dataset = KOADataset(root_dir=test_dir, transform=test_transform)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=2)

print(f"✅ تعداد داده‌های تست پیدا شده: {len(test_dataset)}")

# ---------------------------------------------------------
# ۲. بارگذاری مدل ConvNeXt Ordinal (اصلاح نام لایه به head)
# ---------------------------------------------------------
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

class ConvNeXtOrdinal(nn.Module):
    def __init__(self, num_classes=5):
        super(ConvNeXtOrdinal, self).__init__()
        self.backbone = timm.create_model('convnext_small', pretrained=False, num_classes=0)
        in_features = self.backbone.num_features
        # تغییر fc به head برای انطباق با state_dict مدل شما
        self.head = nn.Linear(in_features, num_classes - 1)

    def forward(self, x):
        features = self.backbone(x)
        logits = self.head(features)
        return logits, features

model_path = "best_ordinal_convnext.pth"
model = ConvNeXtOrdinal(num_classes=5)

checkpoint = torch.load(model_path, map_location=device)
if isinstance(checkpoint, dict) and 'model_state_dict' in checkpoint:
    model.load_state_dict(checkpoint['model_state_dict'])
elif isinstance(checkpoint, dict) and 'state_dict' in checkpoint:
    model.load_state_dict(checkpoint['state_dict'])
else:
    model.load_state_dict(checkpoint)

model.to(device)
model.eval()
print("✅ مدل با موفقیت و بدون خطا بارگذاری شد!")

# ---------------------------------------------------------
# ۳. اجرای ارزیابی و استخراج فیچرها
# ---------------------------------------------------------
all_y_true = []
all_y_probs = []
all_y_preds = []
all_features = []

print("⏳ در حال استخراج خروجی‌ها روی Test Set...")
with torch.no_grad():
    for images, labels in test_loader:
        images = images.to(device)
        logits, feats = model(images)

        # اصلاح باگ AUC روی خروجی‌های Ordinal
        cum_probs = torch.sigmoid(logits).cpu().numpy()

        num_samples = cum_probs.shape[0]
        probs = np.zeros((num_samples, 5))
        probs[:, 0] = 1.0 - cum_probs[:, 0]
        probs[:, 1] = cum_probs[:, 0] - cum_probs[:, 1]
        probs[:, 2] = cum_probs[:, 1] - cum_probs[:, 2]
        probs[:, 3] = cum_probs[:, 2] - cum_probs[:, 3]
        probs[:, 4] = cum_probs[:, 3]

        preds = np.argmax(probs, axis=1)

        all_y_true.extend(labels.numpy())
        all_y_probs.append(probs)
        all_y_preds.extend(preds)
        all_features.append(feats.cpu().numpy())

y_true = np.array(all_y_true)
y_probs = np.vstack(all_y_probs)
y_preds = np.array(all_y_preds)
features = np.vstack(all_features)

# ---------------------------------------------------------
# ۴. محاسبه معیارها و Bootstrap 95% CI
# ---------------------------------------------------------
acc = accuracy_score(y_true, y_preds)
macro_f1 = f1_score(y_true, y_preds, average='macro')
qwk = cohen_kappa_score(y_true, y_preds, weights='quadratic')
mae = mean_absolute_error(y_true, y_preds)
adj_acc = np.mean(np.abs(y_true - y_preds) <= 1)

ovr_aucs = roc_auc_score(y_true, y_probs, multi_class='ovr', average=None)
macro_auc = np.mean(ovr_aucs)

print("\n" + "="*40)
print("📊 نتایج دقیق و اصلاح‌شده روی Test Set:")
print("="*40)
print(f"Accuracy:            {acc*100:.2f}%")
print(f"Macro F1:            {macro_f1:.4f}")
print(f"QWK (Kappa):         {qwk:.4f}")
print(f"MAE:                 {mae:.4f}")
print(f"Adjacent Acc (±1):   {adj_acc*100:.2f}%")
print(f"Corrected Macro-AUC: {macro_auc:.4f}")

# Bootstrap CI
n_iterations = 500
boot_f1s, boot_qwks = [], []
np.random.seed(2026)

for _ in range(n_iterations):
    indices = np.random.choice(len(y_true), size=len(y_true), replace=True)
    if len(np.unique(y_true[indices])) < 5:
        continue
    boot_f1s.append(f1_score(y_true[indices], y_preds[indices], average='macro'))
    boot_qwks.append(cohen_kappa_score(y_true[indices], y_preds[indices], weights='quadratic'))

f1_ci = np.percentile(boot_f1s, [2.5, 97.5])
qwk_ci = np.percentile(boot_qwks, [2.5, 97.5])

print("\n" + "="*40)
print("🛡️ بازه‌های اطمینان آماری (95% CI):")
print("="*40)
print(f"Macro F1 95% CI: [{f1_ci[0]:.4f} - {f1_ci[1]:.4f}]")
print(f"QWK 95% CI:      [{qwk_ci[0]:.4f} - {qwk_ci[1]:.4f}]")

# ---------------------------------------------------------
# ۵. ذخیره فایل‌های سبک برای دانلود
# ---------------------------------------------------------
cm = confusion_matrix(y_true, y_preds)
np.savez_compressed(
    "test_predictions_and_features.npz",
    y_true=y_true,
    y_probs=y_probs,
    y_preds=y_preds,
    features=features,
    confusion_matrix=cm
)

print("\n" + "🎉"*20)
print("✅ خروجی‌ها با موفقیت ذخیره شدند!")
print("فایل test_predictions_and_features.npz آماده دانلود است.")
print("🎉"*20)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
from sklearn.metrics import accuracy_score, f1_score, cohen_kappa_score, mean_absolute_error, roc_auc_score

# ---------------------------------------------------------
# ۱. توابع محاسبه معیارهای ترتیبی و آماری جامع
# ---------------------------------------------------------
def calculate_comprehensive_metrics(y_true, y_pred, y_probs=None):
    acc = accuracy_score(y_true, y_pred)
    macro_f1 = f1_score(y_true, y_pred, average='macro')
    qwk = cohen_kappa_score(y_true, y_pred, weights='quadratic')
    mae = mean_absolute_error(y_true, y_pred)

    # Adjacent Accuracy (اختلاف حداکثر ۱ درجه)
    adj_acc = np.mean(np.abs(y_true - y_pred) <= 1)

    # Macro AUC (در صورت وجود احتمالات)
    macro_auc = None
    if y_probs is not None:
        try:
            # تبدیل به One-Hot برای محاسبه Multi-class ROC AUC
            y_true_onehot = np.eye(5)[y_true]
            macro_auc = roc_auc_score(y_true_onehot, y_probs, average='macro', multi_class='ovr')
        except Exception as e:
            macro_auc = f"N/A ({str(e)})"

    return {
        "Accuracy": acc,
        "Macro F1": macro_f1,
        "QWK": qwk,
        "MAE": mae,
        "Adjacent Accuracy": adj_acc,
        "Macro AUC": macro_auc
    }

# ---------------------------------------------------------
# ۲. تابع استخراج احتمالات و خروجی از مدل Ordinal
# ---------------------------------------------------------
def predict_ordinal_model(model, test_loader, device='cuda'):
    model.eval()
    all_preds = []
    all_probs = []
    all_targets = []

    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs = inputs.to(device)
            logits = model(inputs) # خروجی شامل ۴ لوجیت برای K-1 مرز ترتیبی
            sigmoids = torch.sigmoid(logits)

            # تبدیل لاجیت‌های ترتیبی به احتمالات ۵ کلاس (p_0, p_1, p_2, p_3, p_4)
            # P(Y >= k) = sigmoid(logits_k)
            p_ge1 = sigmoids[:, 0]
            p_ge2 = sigmoids[:, 1]
            p_ge3 = sigmoids[:, 2]
            p_ge4 = sigmoids[:, 3]

            p0 = 1.0 - p_ge1
            p1 = p_ge1 - p_ge2
            p2 = p_ge2 - p_ge3
            p3 = p_ge3 - p_ge4
            p4 = p_ge4

            probs = torch.stack([p0, p1, p2, p3, p4], dim=1)
            # اصلاح مقادیر احتمالات احتمالی منفی به دلیل نویز عددی
            probs = torch.clamp(probs, min=0.0, max=1.0)
            probs = probs / probs.sum(dim=1, keepdim=True)

            preds = (sigmoids > 0.5).sum(dim=1) # کلاس پیش‌بینی‌شده با حد آستانه

            all_probs.append(probs.cpu().numpy())
            all_preds.append(preds.cpu().numpy())
            all_targets.append(targets.numpy())

    return np.vstack(all_targets), np.concatenate(all_preds), np.vstack(all_probs)

# ---------------------------------------------------------
# ۳. تابع استخراج احتمالات از مدل‌های Multi-class معمولی (Focal Loss / ResNet)
# ---------------------------------------------------------
def predict_standard_model(model, test_loader, device='cuda'):
    model.eval()
    all_preds = []
    all_probs = []
    all_targets = []

    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs = inputs.to(device)
            logits = model(inputs)
            probs = F.softmax(logits, dim=1)
            preds = torch.argmax(probs, dim=1)

            all_probs.append(probs.cpu().numpy())
            all_preds.append(preds.cpu().numpy())
            all_targets.append(targets.numpy())

    return np.vstack(all_targets).ravel(), np.concatenate(all_preds), np.vstack(all_probs)

# ---------------------------------------------------------
# ۴. اجرای ارزیابی روی تمام مدل‌ها و ذخیره فایل‌های Prediction
# ---------------------------------------------------------
# فرض بر این است که test_loader تعریف شده است:

print("=== ۱. ارزیابی مدل ConvNeXt Ordinal (Proposed) ===")
y_true_ord, y_pred_ord, y_prob_ord = predict_ordinal_model(ordinal_model, test_loader)
metrics_ord = calculate_comprehensive_metrics(y_true_ord.ravel(), y_pred_ord, y_prob_ord)
for k, v in metrics_ord.items():
    print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")

# ذخیره خروجی‌ها برای تست‌های آماری بعدی
np.savez("predictions_ordinal.npz", y_true=y_true_ord, y_pred=y_pred_ord, y_prob=y_prob_ord)

print("\n=== ۲. ارزیابی مدل ConvNeXt + Focal Loss ===")
try:
    y_true_focal, y_pred_focal, y_prob_focal = predict_standard_model(focal_model, test_loader)
    metrics_focal = calculate_comprehensive_metrics(y_true_focal, y_pred_focal, y_prob_focal)
    for k, v in metrics_focal.items():
        print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")
    np.savez("predictions_focal.npz", y_true=y_true_focal, y_pred=y_pred_focal, y_prob=y_prob_focal)
except Exception as e:
    print(f"خطا در ارزیابی مدل Focal Loss (احتمالاً متغیر focal_model بارگذاری نشده است): {e}")

print("\n=== ۳. ارزیابی مدل Best ResNet Baseline ===")
try:
    y_true_resnet, y_pred_resnet, y_prob_resnet = predict_standard_model(resnet_model, test_loader)
    metrics_resnet = calculate_comprehensive_metrics(y_true_resnet, y_pred_resnet, y_prob_resnet)
    for k, v in metrics_resnet.items():
        print(f"  {k}: {v:.4f}" if isinstance(v, float) else f"  {k}: {v}")
    np.savez("predictions_resnet.npz", y_true=y_true_resnet, y_pred=y_pred_resnet, y_prob=y_prob_resnet)
except Exception as e:
    print(f"خطا در ارزیابی مدل ResNet (احتمالاً متغیر resnet_model بارگذاری نشده است): {e}")
```

```python
def predict_ordinal_model_fixed(model, test_loader, device='cuda'):
    model.eval()
    all_preds = []
    all_probs = []
    all_targets = []

    with torch.no_grad():
        for inputs, targets in test_loader:
            inputs = inputs.to(device)
            logits = model(inputs)
            sigmoids = torch.sigmoid(logits)  # [B, 4]

            # 1. اعمال تصحیح یکنواختی (Monotonic Enforcement)
            # احتمالات نباید افزایش یابند
            sigmoids_np = sigmoids.cpu().numpy()
            for i in range(1, 4):
                sigmoids_np[:, i] = np.minimum(sigmoids_np[:, i], sigmoids_np[:, i-1])

            p_ge1 = sigmoids_np[:, 0]
            p_ge2 = sigmoids_np[:, 1]
            p_ge3 = sigmoids_np[:, 2]
            p_ge4 = sigmoids_np[:, 3]

            p0 = 1.0 - p_ge1
            p1 = p_ge1 - p_ge2
            p2 = p_ge2 - p_ge3
            p3 = p_ge3 - p_ge4
            p4 = p_ge4

            probs = np.stack([p0, p1, p2, p3, p4], axis=1)
            probs = np.clip(probs, 0.0, 1.0)

            # نرمال‌سازی برای اینکه مجموع احتمالات ۱ شود
            sum_p = probs.sum(axis=1, keepdims=True)
            sum_p[sum_p == 0] = 1.0
            probs = probs / sum_p

            # محاسبه کلاس پیش‌بینی‌شده براساس مجموع مرزهای عبور کرده (آستانه 0.5)
            preds = (sigmoids_np > 0.5).sum(axis=1)

            all_probs.append(probs)
            all_preds.append(preds)
            all_targets.append(targets.numpy())

    return np.concatenate(all_targets).ravel(), np.concatenate(all_preds), np.concatenate(all_probs, axis=0)
```

```python
import numpy as np
import torch
import torch.nn.functional as F
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score, cohen_kappa_score, confusion_matrix

def evaluate_ordinal_model(model, dataloader, device='cuda', num_classes=4):
    model.eval()

    all_targets = []
    all_pred_classes = []
    all_probs = []

    with torch.no_grad():
        for images, targets in dataloader:
            images = images.to(device)
            targets = targets.to(device)

            logits = model(images)

            # اگر خروجی مدل ۳ تایی باشد (Ordinal thresholds) یا ۴ تایی
            if logits.shape[1] == num_classes - 1:
                sigmoids = torch.sigmoid(logits)
                p0 = 1 - sigmoids[:, 0]
                p1 = sigmoids[:, 0] - sigmoids[:, 1]
                p2 = sigmoids[:, 1] - sigmoids[:, 2]
                p3 = sigmoids[:, 2]
                probs = torch.stack([p0, p1, p2, p3], dim=1)
                probs = torch.clamp(probs, min=0.0, max=1.0)
                preds = (sigmoids > 0.5).cumprod(dim=1).sum(dim=1)
            else:
                # اگر خروجی ۴ کلاسه استاندارد باشد
                probs = F.softmax(logits, dim=1)
                preds = torch.argmax(probs, dim=1)

            all_targets.extend(targets.cpu().numpy())
            all_pred_classes.extend(preds.cpu().numpy())
            all_probs.extend(probs.cpu().numpy())

    all_targets = np.array(all_targets)
    all_pred_classes = np.array(all_pred_classes)
    all_probs = np.array(all_probs)

    # ۱. نرمال‌سازی قطعی تا جمع هر سطر دقیقاً ۱.۰ شود
    row_sums = all_probs.sum(axis=1, keepdims=True)
    row_sums[row_sums == 0] = 1.0  # جلوگیری از تقسیم بر صفر
    all_probs = all_probs / row_sums

    # ۲. محاسبه معیارهای ارزیابی اصلی
    acc = accuracy_score(all_targets, all_pred_classes)
    f1_macro = f1_score(all_targets, all_pred_classes, average='macro', zero_division=0)
    qwk = cohen_kappa_score(all_targets, all_pred_classes, weights='quadratic')

    # ۳. محاسبه ROC-AUC با صراحت دادن به برچسب‌های موجود (labels)
    possible_labels = list(range(num_classes))
    try:
        roc_auc = roc_auc_score(
            all_targets,
            all_probs,
            multi_class='ovr',
            average='macro',
            labels=possible_labels
        )
    except Exception as e:
        print(f"⚠️ هشدار در محاسبه ROC-AUC: {e}")
        roc_auc = 0.0

    cm = confusion_matrix(all_targets, all_pred_classes, labels=possible_labels)

    # نمایش نتایج
    print("=" * 45)
    print("📊 نتایج ارزیابی مدل Ordinal (ConvNeXt-Small)")
    print("=" * 45)
    print(f"🌟 ROC-AUC (Macro OVR): {roc_auc * 100:.2f}%")
    print(f"🎯 Accuracy           : {acc * 100:.2f}%")
    print(f"📈 F1-Score (Macro)   : {f1_macro * 100:.2f}%")
    print(f"⚖️️ Quadratic Kappa (QWK): {qwk:.4f}")
    print("=" * 45)
    print("📉 Confusion Matrix:")
    print(cm)
    print("=" * 45)

    return roc_auc, acc, f1_macro, qwk

# اجرای ارزیابی
roc_auc, acc, f1, qwk = evaluate_ordinal_model(ordinal_model, test_loader, device=device)
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader
import timm
import numpy as np
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score, cohen_kappa_score, confusion_matrix, mean_absolute_error

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# ==========================================
# ۱. ساختار مدل Ordinal ConvNeXt
# ==========================================
def create_ordinal_convnext(num_classes=4, pretrained=False):
    # برای N کلاس، N-1 لوگیت آستانه‌ای داریم (برای 4 کلاس -> 3 خروجی)
    model = timm.create_model('convnext_small', pretrained=pretrained, num_classes=num_classes - 1)
    return model

def load_trained_model(weights_path='best_ordinal_convnext.pth', num_classes=4):
    ordinal_model = create_ordinal_convnext(num_classes=num_classes, pretrained=False)

    checkpoint = torch.load(weights_path, map_location=device, weights_only=False)
    state_dict = checkpoint.get('state_dict', checkpoint)

    clean_state_dict = {}
    for k, v in state_dict.items():
        name = k.replace("backbone.", "").replace("module.", "")
        clean_state_dict[name] = v

    if 'head.weight' in clean_state_dict and hasattr(ordinal_model, 'head') and hasattr(ordinal_model.head, 'fc'):
        clean_state_dict['head.fc.weight'] = clean_state_dict.pop('head.weight')
        clean_state_dict['head.fc.bias'] = clean_state_dict.pop('head.bias')

    missing, unexpected = ordinal_model.load_state_dict(clean_state_dict, strict=False)
    ordinal_model = ordinal_model.to(device)
    ordinal_model.eval()

    print(f"✅ مدل با موفقیت لود شد. (Missing: {len(missing)}, Unexpected: {len(unexpected)})")
    return ordinal_model

# ==========================================
# ۲. تابع ارزیابی دقیق و اصلاح‌شده
# ==========================================
def evaluate_ordinal_correct(model, dataloader, device='cuda', num_classes=4):
    """
    محاسبه دقیق احتمالات کلاس‌ها، اصلاح باگ AUC و افزودن معیارهای Ordinal
    """
    model.eval()
    all_targets = []
    all_preds = []
    all_class_probs = []

    with torch.no_grad():
        for images, targets in dataloader:
            images = images.to(device)
            targets = targets.to(device)

            # ۱. لوگیت‌های آستانه: [batch_size, num_classes - 1]
            logits = model(images)
            probs_gt = torch.sigmoid(logits)  # احتمال P(Y > k) برای k=0, 1, ..., num_classes-2

            # ۲. پیش‌بینی برچسب سخت بر اساس تعداد آستانه‌های رد شده (> 0.5)
            preds = (probs_gt > 0.5).sum(dim=1)

            # ۳. نگاشت درست و ریاضی P(Y > k) به P(Y = k)
            # P(Y = 0) = 1 - P(Y > 0)
            # P(Y = k) = P(Y > k-1) - P(Y > k)
            # P(Y = N-1) = P(Y > N-2)
            batch_size = probs_gt.shape[0]
            class_probs = torch.zeros((batch_size, num_classes), device=device)

            class_probs[:, 0] = 1.0 - probs_gt[:, 0]
            for k in range(1, num_classes - 1):
                class_probs[:, k] = probs_gt[:, k - 1] - probs_gt[:, k]
            class_probs[:, num_classes - 1] = probs_gt[:, num_classes - 2]

            # جلوگیری از اعداد منفی احتمالی ناشی از عدم یکنوایی ناچیز لوگیت‌ها
            class_probs = torch.clamp(class_probs, min=1e-7, max=1.0)
            # نرمال‌سازی مجموع احتمالات برابر ۱ (Sum-to-One Normalization)
            class_probs = class_probs / class_probs.sum(dim=1, keepdim=True)

            all_targets.extend(targets.cpu().numpy())
            all_preds.extend(preds.cpu().numpy())
            all_class_probs.extend(class_probs.cpu().numpy())

    all_targets = np.array(all_targets)
    all_preds = np.array(all_preds)
    all_class_probs = np.array(all_class_probs)

    # --- محاسبه معیارها ---
    acc = accuracy_score(all_targets, all_preds)
    f1_macro = f1_score(all_targets, all_preds, average='macro', zero_division=0)
    qwk = cohen_kappa_score(all_targets, all_preds, weights='quadratic')
    mae = mean_absolute_error(all_targets, all_preds)

    # Adjacent Accuracy (خطای حداکثر ۱ درجه)
    adjacent_acc = np.mean(np.abs(all_targets - all_preds) <= 1)

    # محاسبه صحیح Macro AUC (One-vs-Rest)
    present_labels = np.unique(all_targets)
    if len(present_labels) == num_classes:
        roc_auc = roc_auc_score(all_targets, all_class_probs, multi_class='ovr', average='macro')
    else:
        filtered_probs = all_class_probs[:, present_labels]
        filtered_probs = filtered_probs / filtered_probs.sum(axis=1, keepdims=True)
        roc_auc = roc_auc_score(all_targets, filtered_probs, multi_class='ovr', average='macro', labels=present_labels)

    cm = confusion_matrix(all_targets, all_preds, labels=list(range(num_classes)))

    print("=" * 65)
    print("📊 نتایج دقیق و اصلاح‌شده ارزیابی مدل Ordinal (ConvNeXt-Small)")
    print("=" * 65)
    print(f"🌟 Macro ROC-AUC       : {roc_auc * 100:.2f}% (اصلاح باگ)")
    print(f"🎯 Overall Accuracy    : {acc * 100:.2f}%")
    print(f"📈 Macro F1-Score      : {f1_macro * 100:.2f}%")
    print(f"⚖️ Quadratic Kappa (QWK): {qwk:.4f}")
    print(f"📐 Mean Absolute Error : {mae:.4f}")
    print(f"✅ Adjacent Acc (±1)   : {adjacent_acc * 100:.2f}%")
    print("=" * 65)
    print("📉 Confusion Matrix:")
    print(cm)
    print("=" * 65)

    return {
        'roc_auc': roc_auc,
        'acc': acc,
        'f1_macro': f1_macro,
        'qwk': qwk,
        'mae': mae,
        'adjacent_acc': adjacent_acc,
        'cm': cm
    }
```

```python
import os
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image
import timm
import numpy as np
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score, cohen_kappa_score, confusion_matrix, mean_absolute_error

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# ==========================================
# ۱. دانلود و دریافت مسیر دیتاست از کگل
# ==========================================
print("⏳ در حال دریافت مسیر دیتاست از کگل...")
dataset_path = get_dataset_path()
print("📂 مسیر دیتاست:", dataset_path)

# پیدا کردن پوشه تست (معمولاً auto-test یا test داخل پوشه‌های دیتاست است)
# ما مستقیماً پوشه test اصلی را لود می‌کنیم (نه auto-test)
test_dir = None
for root, dirs, files in os.walk(dataset_path):
    if os.path.basename(root).lower() == 'test' and 'auto' not in root.lower():
        test_dir = root
        break

if test_dir is None:
    # ساخت مسیر پیش‌فرض بر اساس ساختار رایج این دیتاست
    test_dir = os.path.join(dataset_path, "kneeKL224", "test")
    if not os.path.exists(test_dir):
        test_dir = os.path.join(dataset_path, "test")

print(f"🎯 پوشه ارزیابی تست: {test_dir}")

# ==========================================
# ۲. استخراج مسیر عکس‌ها و برچسب‌های ۵ کلاس
# ==========================================
image_paths = []
labels = []

# کلاس‌های 0 تا 4
for label in range(5):
    # بررسی پوشه‌های نام‌گذاری شده مثل '0' یا '0_Normal'
    class_folder = None
    if os.path.exists(test_dir):
        for d in os.listdir(test_dir):
            if d == str(label) or d.startswith(f"{label}_"):
                class_folder = os.path.join(test_dir, d)
                break

    if class_folder and os.path.exists(class_folder):
        for img_name in os.listdir(class_folder):
            if img_name.lower().endswith(('.png', '.jpg', '.jpeg')):
                image_paths.append(os.path.join(class_folder, img_name))
                labels.append(label)

print(f"✅ تعداد کل تصاویر تست پیدا شده: {len(image_paths)} (انتظار: 1656)")

# ==========================================
# ۳. ساخت کلاس Dataset و DataLoader
# ==========================================
class KneeTestDataset(Dataset):
    def __init__(self, image_paths, labels, transform=None):
        self.image_paths = image_paths
        self.labels = labels
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        image = Image.open(img_path).convert('RGB')
        label = self.labels[idx]

        if self.transform:
            image = self.transform(image)

        return image, label

test_transform = transforms.Compose([
    transforms.Resize((448, 448)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

test_dataset = KneeTestDataset(image_paths, labels, transform=test_transform)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=2)

# ==========================================
# ۴. تعریف مدل و تابع بارگذاری وزن‌ها
# ==========================================
def create_ordinal_convnext(num_classes=5, pretrained=False):
    # برای 5 کلاس، 4 خروجی آستانه تعریف می‌شود
    model = timm.create_model('convnext_small', pretrained=pretrained, num_classes=num_classes - 1)
    return model

def load_trained_model(weights_path='best_ordinal_convnext.pth', num_classes=5):
    ordinal_model = create_ordinal_convnext(num_classes=num_classes, pretrained=False)

    if not os.path.exists(weights_path):
        raise FileNotFoundError(f"❌ فایل وزن‌های {weights_path} یافت نشد! لطفاً فایل pth را در مسیر جاری قرار دهید.")

    checkpoint = torch.load(weights_path, map_location=device, weights_only=False)
    state_dict = checkpoint.get('state_dict', checkpoint)

    clean_state_dict = {}
    for k, v in state_dict.items():
        name = k.replace("backbone.", "").replace("module.", "")
        clean_state_dict[name] = v

    if 'head.weight' in clean_state_dict and hasattr(ordinal_model, 'head') and hasattr(ordinal_model.head, 'fc'):
        clean_state_dict['head.fc.weight'] = clean_state_dict.pop('head.weight')
        clean_state_dict['head.fc.bias'] = clean_state_dict.pop('head.bias')

    missing, unexpected = ordinal_model.load_state_dict(clean_state_dict, strict=False)
    ordinal_model = ordinal_model.to(device)
    ordinal_model.eval()

    print(f"✅ مدل با موفقیت لود شد.")
    return ordinal_model

# ==========================================
# ۵. تابع ارزیابی دقیق (اصلاح AUC و معیارهای Ordinal)
# ==========================================
def evaluate_ordinal_correct(model, dataloader, device='cuda', num_classes=5):
    model.eval()
    all_targets = []
    all_preds = []
    all_class_probs = []

    with torch.no_grad():
        for images, targets in dataloader:
            images = images.to(device)
            targets = targets.to(device)

            # ۱. لوگیت‌های آستانه: [batch_size, 4]
            logits = model(images)
            probs_gt = torch.sigmoid(logits)  # P(Y > k)

            # ۲. پیش‌بینی برچسب بر اساس آستانه‌ها
            preds = (probs_gt > 0.5).sum(dim=1)

            # ۳. محاسبه صحیح احتمالات ۵ کلاس P(Y = k)
            batch_size = probs_gt.shape[0]
            class_probs = torch.zeros((batch_size, num_classes), device=device)

            class_probs[:, 0] = 1.0 - probs_gt[:, 0]
            for k in range(1, num_classes - 1):
                class_probs[:, k] = probs_gt[:, k - 1] - probs_gt[:, k]
            class_probs[:, num_classes - 1] = probs_gt[:, num_classes - 2]

            class_probs = torch.clamp(class_probs, min=1e-7, max=1.0)
            class_probs = class_probs / class_probs.sum(dim=1, keepdim=True)

            all_targets.extend(targets.cpu().numpy())
            all_preds.extend(preds.cpu().numpy())
            all_class_probs.extend(class_probs.cpu().numpy())

    all_targets = np.array(all_targets)
    all_preds = np.array(all_preds)
    all_class_probs = np.array(all_class_probs)

    # محاسبه معیارهای تخصصی
    acc = accuracy_score(all_targets, all_preds)
    f1_macro = f1_score(all_targets, all_preds, average='macro', zero_division=0)
    qwk = cohen_kappa_score(all_targets, all_preds, weights='quadratic')
    mae = mean_absolute_error(all_targets, all_preds)
    adjacent_acc = np.mean(np.abs(all_targets - all_preds) <= 1)

    # محاسبه صحیح Macro AUC
    present_labels = np.unique(all_targets)
    if len(present_labels) == num_classes:
        roc_auc = roc_auc_score(all_targets, all_class_probs, multi_class='ovr', average='macro')
    else:
        filtered_probs = all_class_probs[:, present_labels]
        filtered_probs = filtered_probs / filtered_probs.sum(axis=1, keepdims=True)
        roc_auc = roc_auc_score(all_targets, filtered_probs, multi_class='ovr', average='macro', labels=present_labels)

    cm = confusion_matrix(all_targets, all_preds, labels=list(range(num_classes)))

    print("\n" + "=" * 65)
    print("📊 نتایج دقیق و اصلاح‌شده ارزیابی مدل Ordinal (ConvNeXt-Small)")
    print("=" * 65)
    print(f"🌟 Macro ROC-AUC       : {roc_auc * 100:.2f}%")
    print(f"🎯 Overall Accuracy    : {acc * 100:.2f}%")
    print(f"📈 Macro F1-Score      : {f1_macro * 100:.2f}%")
    print(f"⚖️ Quadratic Kappa (QWK): {qwk:.4f}")
    print(f"📐 Mean Absolute Error : {mae:.4f}")
    print(f"✅ Adjacent Acc (±1)   : {adjacent_acc * 100:.2f}%")
    print("=" * 65)
    print("📉 Confusion Matrix (KL Grades 0 to 4):")
    print(cm)
    print("=" * 65)

# ==========================================
# ۶. اجرای ارزیابی
# ==========================================
if __name__ == '__main__':
    # حتما فایل best_ordinal_convnext.pth کنار این فایل پایتون باشد
    model = load_trained_model(weights_path='best_ordinal_convnext.pth', num_classes=5)
    evaluate_ordinal_correct(model, test_loader, device=device, num_classes=5)
```

```python
import os
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from torchvision import transforms, models
import numpy as np
from sklearn.metrics import roc_auc_score, accuracy_score, f1_score, cohen_kappa_score, confusion_matrix, mean_absolute_error

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# ==========================================
# ۱. تابع لود کردن مدل ResNet18 (اصلاح‌شده)
# ==========================================
def load_resnet18_model(weights_path='best_resnet18_448_seed2026.pth', num_classes=5):
    # بارگذاری فایل وزن‌ها
    checkpoint = torch.load(weights_path, map_location=device, weights_only=False)
    state_dict = checkpoint.get('state_dict', checkpoint)

    # پاک‌سازی پیشوندهای احتمالی
    clean_state_dict = {}
    for k, v in state_dict.items():
        name = k.replace("backbone.", "").replace("module.", "").replace("model.", "")
        clean_state_dict[name] = v

    # ساخت معماری بدون پارامتر اضافه
    model = models.resnet18(weights=None)

    # تنظیم لایه fc با توجه به وزن‌های ذخیره شده
    in_features = model.fc.in_features
    if 'fc.weight' in clean_state_dict:
        out_features = clean_state_dict['fc.weight'].shape[0]
        model.fc = nn.Linear(in_features, out_features)
    else:
        model.fc = nn.Linear(in_features, num_classes)

    missing, unexpected = model.load_state_dict(clean_state_dict, strict=False)
    model = model.to(device)
    model.eval()

    print(f"✅ مدل ResNet18 با موفقیت لود شد. (Missing: {len(missing)}, Unexpected: {len(unexpected)})")
    return model

# ==========================================
# ۲. تابع ارزیابی مدل
# ==========================================
def evaluate_standard_model(model, dataloader, device='cuda', num_classes=5):
    model.eval()
    all_targets = []
    all_preds = []
    all_class_probs = []

    with torch.no_grad():
        for images, targets in dataloader:
            images = images.to(device)
            targets = targets.to(device)

            logits = model(images)

            # خروجی استاندارد ۵ کلاس
            if logits.shape[1] == num_classes:
                probs = torch.softmax(logits, dim=1)
                preds = torch.argmax(probs, dim=1)
            # خروجی آستانه‌ای (Ordinal)
            elif logits.shape[1] == num_classes - 1:
                probs_gt = torch.sigmoid(logits)
                preds = (probs_gt > 0.5).sum(dim=1)

                batch_size = probs_gt.shape[0]
                probs = torch.zeros((batch_size, num_classes), device=device)
                probs[:, 0] = 1.0 - probs_gt[:, 0]
                for k in range(1, num_classes - 1):
                    probs[:, k] = probs_gt[:, k - 1] - probs_gt[:, k]
                probs[:, num_classes - 1] = probs_gt[:, num_classes - 2]
                probs = torch.clamp(probs, min=1e-7, max=1.0)
                probs = probs / probs.sum(dim=1, keepdim=True)

            all_targets.extend(targets.cpu().numpy())
            all_preds.extend(preds.cpu().numpy())
            all_class_probs.extend(probs.cpu().numpy())

    all_targets = np.array(all_targets)
    all_preds = np.array(all_preds)
    all_class_probs = np.array(all_class_probs)

    # محاسبه معیارها
    acc = accuracy_score(all_targets, all_preds)
    f1_macro = f1_score(all_targets, all_preds, average='macro', zero_division=0)
    qwk = cohen_kappa_score(all_targets, all_preds, weights='quadratic')
    mae = mean_absolute_error(all_targets, all_preds)
    adjacent_acc = np.mean(np.abs(all_targets - all_preds) <= 1)

    # محاسبه Macro AUC
    present_labels = np.unique(all_targets)
    if len(present_labels) == num_classes:
        roc_auc = roc_auc_score(all_targets, all_class_probs, multi_class='ovr', average='macro')
    else:
        filtered_probs = all_class_probs[:, present_labels]
        filtered_probs = filtered_probs / filtered_probs.sum(axis=1, keepdims=True)
        roc_auc = roc_auc_score(all_targets, filtered_probs, multi_class='ovr', average='macro', labels=present_labels)

    cm = confusion_matrix(all_targets, all_preds, labels=list(range(num_classes)))

    print("\n" + "=" * 65)
    print("📊 نتایج دقیق ارزیابی مدل ResNet18 (448x448)")
    print("=" * 65)
    print(f"🌟 Macro ROC-AUC       : {roc_auc * 100:.2f}%")
    print(f"🎯 Overall Accuracy    : {acc * 100:.2f}%")
    print(f"📈 Macro F1-Score      : {f1_macro * 100:.2f}%")
    print(f"⚖️ Quadratic Kappa (QWK): {qwk:.4f}")
    print(f"📐 Mean Absolute Error : {mae:.4f}")
    print(f"✅ Adjacent Acc (±1)   : {adjacent_acc * 100:.2f}%")
    print("=" * 65)
    print("📉 Confusion Matrix:")
    print(cm)
    print("=" * 65)

# ==========================================
# ۳. اجرای ارزیابی
# ==========================================
if __name__ == '__main__':
    resnet_model = load_resnet18_model('best_resnet18_448_seed2026.pth', num_classes=5)
    evaluate_standard_model(resnet_model, test_loader, device=device, num_classes=5)
```

```python
import torch
import torch.nn as nn
import timm
from torchvision import models

device = 'cuda' if torch.cuda.is_available() else 'cpu'

# ==========================================
# ۱. تابع بارگذاری مدل EfficientNet-B0
# ==========================================
def inspect_and_load_efficientnet(weights_path='best_efficientnetb0_448_seed2026.pth', num_classes=5):
    checkpoint = torch.load(weights_path, map_location=device, weights_only=False)

    # استخراج state_dict اصلی
    if isinstance(checkpoint, dict):
        if 'state_dict' in checkpoint:
            state_dict = checkpoint['state_dict']
        elif 'model' in checkpoint:
            state_dict = checkpoint['model']
        elif 'model_state_dict' in checkpoint:
            state_dict = checkpoint['model_state_dict']
        else:
            state_dict = checkpoint
    else:
        state_dict = checkpoint

    # تمیزکاری پیشوندهای احتمالی
    new_state_dict = {}
    for k, v in state_dict.items():
        name = k
        for prefix in ["backbone.", "model.", "encoder.", "_orig_mod.", "module."]:
            if name.startswith(prefix):
                name = name[len(prefix):]
        new_state_dict[name] = v

    # تلاش برای ساخت مدل با timm یا torchvision
    try:
        model = timm.create_model('efficientnet_b0', pretrained=False, num_classes=num_classes)
    except Exception:
        model = models.efficientnet_b0(weights=None)
        in_features = model.classifier[1].in_features
        model.classifier[1] = nn.Linear(in_features, num_classes)

    # انطباق لایه آخر در صورت متفاوت بودن ابعاد
    for fc_name in ['classifier.weight', 'classifier.1.weight', 'head.fc.weight', 'fc.weight']:
        if fc_name in new_state_dict:
            out_shape = new_state_dict[fc_name].shape[0]
            if hasattr(model, 'classifier') and isinstance(model.classifier, nn.Sequential):
                in_feat = model.classifier[1].in_features
                model.classifier[1] = nn.Linear(in_feat, out_shape)
            elif hasattr(model, 'classifier') and isinstance(model.classifier, nn.Linear):
                in_feat = model.classifier.in_features
                model.classifier = nn.Linear(in_feat, out_shape)
            elif hasattr(model, 'reset_classifier'):
                model.reset_classifier(num_classes=out_shape)
            break

    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)
    model = model.to(device)
    model.eval()

    print("--------------------------------------------------")
    print(f"✅ وضعیت بارگذاری EfficientNet-B0:")
    print(f"❌ Missing Keys   : {len(missing)}")
    print(f"❓ Unexpected Keys: {len(unexpected)}")
    print("--------------------------------------------------")

    return model

# ==========================================
# ۲. اجرای ارزیابی
# ==========================================
if __name__ == '__main__':
    effnet_model = inspect_and_load_efficientnet('best_efficientnetb0_448_seed2026.pth', num_classes=5)
    evaluate_standard_model(effnet_model, test_loader, device=device, num_classes=5)
```

```python
import torch
import torch.nn as nn
import timm
from torchvision import models

device = 'cuda' if torch.cuda.is_available() else 'cpu'

def inspect_and_load_efficientnet_smart(weights_path='best_efficientnetb0_448_seed2026.pth', num_classes=5):
    checkpoint = torch.load(weights_path, map_location=device, weights_only=False)

    # استخراج state_dict اصلی
    if isinstance(checkpoint, dict):
        if 'state_dict' in checkpoint:
            state_dict = checkpoint['state_dict']
        elif 'model' in checkpoint:
            state_dict = checkpoint['model']
        elif 'model_state_dict' in checkpoint:
            state_dict = checkpoint['model_state_dict']
        else:
            state_dict = checkpoint
    else:
        state_dict = checkpoint

    # تمیزکاری پیشوندها
    cleaned_dict = {}
    for k, v in state_dict.items():
        name = k
        for prefix in ["backbone.", "model.", "encoder.", "_orig_mod.", "module."]:
            if name.startswith(prefix):
                name = name[len(prefix):]
        cleaned_dict[name] = v

    print(f"📌 ۵ کلید اول فایل weights برای EfficientNet:")
    for i, k in enumerate(list(cleaned_dict.keys())[:5]):
        print(f"   {i+1}. {k}")

    # تشخیص اینکه آیا از timm استفاده شده یا torchvision
    is_timm = any(k.startswith('conv_stem') or k.startswith('blocks') for k in cleaned_dict.keys())

    if is_timm:
        print("💡 معماری تشخیص داده شده: timm (efficientnet_b0)")
        model = timm.create_model('efficientnet_b0', pretrained=False, num_classes=num_classes)
        # انطباق لایه آخر
        if 'classifier.weight' in cleaned_dict:
            out_dim = cleaned_dict['classifier.weight'].shape[0]
            model.reset_classifier(num_classes=out_dim)
    else:
        print("💡 معماری تشخیص داده شده: torchvision (efficientnet_b0)")
        model = models.efficientnet_b0(weights=None)
        in_features = model.classifier[1].in_features
        if 'classifier.1.weight' in cleaned_dict:
            out_dim = cleaned_dict['classifier.1.weight'].shape[0]
            model.classifier[1] = nn.Linear(in_features, out_dim)
        else:
            model.classifier[1] = nn.Linear(in_features, num_classes)

    missing, unexpected = model.load_state_dict(cleaned_dict, strict=False)
    model = model.to(device)
    model.eval()

    print("--------------------------------------------------")
    print(f"✅ وضعیت بارگذاری پس از تطبیق معماری:")
    print(f"❌ Missing Keys   : {len(missing)}")
    print(f"❓ Unexpected Keys: {len(unexpected)}")
    print("--------------------------------------------------")

    return model

# اجرای تست و ارزیابی
eff_model = inspect_and_load_efficientnet_smart('best_efficientnetb0_448_seed2026.pth', num_classes=5)
evaluate_standard_model(eff_model, test_loader, device=device, num_classes=5)
```
